# GeoGuard: Satellite Landslide Detection
## Notebook 03: CNN Model Training & Validation

```python
import os, sys
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.train import run_training
from src.model import build_landslide_cnn

print(f"TensorFlow Version: {tf.__version__}")
```

### 2. Inspecting the Model Architecture

```python
model = build_landslide_cnn(input_shape=(128, 128, 4))
model.summary()
print(f"Total trainable parameters: {model.count_params():,}")
```

### 3. Executing Model Training

```python
data_dir = os.path.join(PROJECT_ROOT, 'data', 'raw')
trained_model, metadata = run_training(
    data_dir=data_dir,
    epochs=25,
    batch_size=16,
    learning_rate=1e-3,
    seed=42
)
print(f"Best Val Loss: {metadata['best_val_loss']:.4f}")
print(f"Best Val Accuracy: {metadata['best_val_accuracy']:.4f}")
```

### 4. Plotting Loss and Accuracy Learning Curves

```python
hist = metadata['history']
epochs_range = range(1, len(hist['loss']) + 1)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(epochs_range, hist['loss'], 'b-o', label='Training Loss')
ax1.plot(epochs_range, hist['val_loss'], 'r--s', label='Validation Loss')
ax1.set_title('Training & Validation Loss', fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()
ax2.plot(epochs_range, hist['accuracy'], 'b-o', label='Training Accuracy')
ax2.plot(epochs_range, hist['val_accuracy'], 'r--s', label='Validation Accuracy')
ax2.set_title('Training & Validation Accuracy', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()
plt.tight_layout()
plt.show()
```
